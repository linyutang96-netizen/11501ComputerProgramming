## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？因為 Python 的 int 可以依照需要自動增加儲存空間，所以能表示非常大的整數
2. Python 的 `int` 是否具有固定的位元數限制？否，不會像其他語言常見的 32 位元或 64 位元整數
3. 什麼是 Arbitrary Precision（任意精度）？意思是整數可以隨著數字變大而動態擴充，不受固定位元數限制。
4. CPython 使用什麼物件來實作 `int`？使用 PyLongObject。
5. Python 整數大小真正的限制是什麼？真正的限制是電腦的記憶體（RAM）大小。只要記憶體夠，就能儲存更大的整數。

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？因為電腦使用二進位儲存浮點數，像 0.1 這種數值無法被完全精確表示，所以會產生微小誤差。
2. 浮點數是否能精確表示所有小數？否，像是
0.1
0.2
0.3
都無法被精確儲存。
3. 什麼是浮點數的精度（precision）？精度是指數字的準確程度，也就是能保留多少位有效數字。
4. 什麼是浮點數的範圍（range）？範圍是指能表示的最大值和最小值。
5. 精度與範圍有何不同？精度：數字有多準確；範圍：數字可以多大或多小

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？FP32、FP16、BF16
2. 什麼是 FP64？64 位元浮點數（雙精度浮點數）。
3. 什麼是 FP32？32 位元浮點數（單精度浮點數）。
4. 什麼是 FP16？16 位元浮點數（半精度浮點數）。
5. 什麼是 BF16？Brain Floating Point 16 的縮寫，是一種 16 位元浮點數格式。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？因為佔用太多記憶體、計算速度較慢、GPU 不適合做這類運算、AI 權重通常是小數，不是整數

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？因為它可以自動增加記憶體來存放更多位數。
2. 為什麼 `float` 卻存在範圍限制？因為 float 使用固定的 64 位元空間來儲存數值。
3. 為什麼 `float` 可能產生誤差？因為很多十進位小數無法用有限的二進位位元精確表示。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。Python int：追求正確性；FP16、BF16：追求速度與效率
2. 各自的優點與缺點。int優點：非常精確 缺點：較耗記憶體、計算較慢；FP16、BF16優點：記憶體較省、計算較快 缺點：會有誤差
3. 為什麼大型語言模型通常選擇較低精度的表示方式？因為可以省記憶體、提升訓練速度和 GPU 運算效率。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？記憶體需求暴增、訓練速度變慢、GPU 效率降低、成本提高
