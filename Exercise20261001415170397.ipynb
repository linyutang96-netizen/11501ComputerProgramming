1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
#### 計算過程

將小數部分持續乘以 2，並記錄每次結果的整數部分。

- 0.1 × 2 = 0.2 → 記錄 0
- 0.2 × 2 = 0.4 → 記錄 0
- 0.4 × 2 = 0.8 → 記錄 0
- 0.8 × 2 = 1.6 → 記錄 1
- 0.6 × 2 = 1.2 → 記錄 1
- 0.2 × 2 = 0.4 → 記錄 0
- 0.4 × 2 = 0.8 → 記錄 0
- 0.8 × 2 = 1.6 → 記錄 1
- 0.6 × 2 = 1.2 → 記錄 1
- ...

#### 計算結果

將記錄的數字依序排列：

`0 → 0 → 0 → 1 → 1 → 0 → 0 → 1 → 1 → ...`

因此：

`0.1₁₀ = 0.000110011001100110011...₂`

#### 理由

在計算過程中會出現：

`0.2 → 0.4 → 0.8 → 1.6 → 0.6 → 1.2 → 0.2`

可以發現又回到 0.2，因此後面的計算會一直重複。

#### 看到的現象

二進位小數部分出現：

`0011 0011 0011 0011 ...`

其中 **0011** 會無限循環重複，因此：

`0.1₁₀ = 0.000110011001100110011...₂`

是一個無限循環小數。

#### 結論

由於 0.1 在二進位中是無限循環小數，因此電腦無法完全精確儲存，只能儲存最接近的近似值，這也是 Python 浮點數誤差的主要原因。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙

Python因為儲存空間有限，無法記錄無限多位數，會先保留有限位數，然後對後面的數字進行近似，並採用「四捨六入五成雙」的方式做捨入，所以Python儲存的不是完全精確的 0.1，而是最接近 0.1 的近似值。

3. Python如何知道浮點數的精度？請問執行以下程式碼:
```python
import sys
print(sys.float_info)
```
並說明輸出結果的意思。
```python
import sys
print(sys.float_info(
max=1.7976931348623157e+308,
min=2.2250738585072014e-308,
dig=15,
mant_dig=53,
epsilon=2.220446049250313e-16,
...)
```

max：float 能表示的最大數值，大約是 1.79 × 10³⁰⁸；min：float 能表示的最小正數，大約是 2.22 × 10⁻³⁰⁸；dig = 15：表示 float 大約可以保留 15 位有效數字；mant_dig = 53：表示浮點數使用 53 個二進位位元來儲存有效數字；epsilon = 2.220446049250313e-16：表示 1 和下一個可表示浮點數之間最小的差距，也就是 float 的精度。

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。
我認為要看使用的情況，在科學、工程或 AI 計算中，感覺使用 float 通常就很夠用了，因為計算速度快，而且誤差非常小。但在金融、銀行或會計系統中，金額必須非常精確，比較適合使用 Decimal，可以避免因為浮點數誤差造成金額計算錯誤。因此，我認為科學計算比較適合使用 float，金融計算則比較適合使用 Decimal。
